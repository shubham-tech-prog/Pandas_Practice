Merging 2 dataframes 

```python
employees = pd.DataFrame({
    "employees_id" : [1, 2, 3, 4, 5],
    "name" : ["John", "Anna", "Peter", "Linda", "Bob"],
    "department" : ["HR","IT","Finance","IT","HR"],
})

# DataFrame 2 : Salary information 
salaries = pd.DataFrame({
    "employees_id" : [1, 2, 3, 6, 7],
    "salary" : [60000, 80000, 65000, 70000, 90000],
    "bonus" : [5000, 10000, 7000, 8000, 12000],
    
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries,on="employees_id",how ="inner")  # It is use to merge 2 dataFrames

                                                            # on = "employee_id" -----> it is use to merge two dataFrames on common coloumn = employee_id 
                                                             #  Basis of Row
```

```python
pd.merge(employees,salaries,on="employees_id",how ="outer") 

# outer --- > ON THE BASIS OF COLOUMN 
```

```python
pd.merge(employees,salaries,on="employees_id",how ="left") 
```

```python
pd.merge(employees,salaries,on="employees_id",how ="right")
```

# Group_By_Aggregation 

```python
import numpy as np 
import pandas as pd
```

```python
data = {
    "Category" : ["A","B","A","B","A","B","A","B"],
    "Store" : ["S1","S1","S2","S2","S1","S2","S2","S1"],
    "Sales" : [100, 200, 150, 250, 120, 180, 200, 300],
    "Date"  : pd.date_range("2023-01-01", periods = 8)
}
df = pd.DataFrame(data)
```

```python
# Group by Category and calculatate the sum of sales 
cat = df.groupby("Category")["Sales"].sum()
cat 

# Sum of individual A , B sales 
```

```python
# Group by store and calculation the sum of sales 
cat = df.groupby("Store")["Sales"].sum()
cat
```

```python
# Group by multiple coloumns 
# Group by category and store
cat = df.groupby(["Category","Store"])["Sales"].sum()
cat
```

```python
# In Aggregation there are , Mean , median , Std , count   .... etc 
```

```python
df["Sales"].mean()   # Mean
```

```python
df
```

```python
df["Sales"].agg(["sum","mean","min","max","count","std","median"])

# agg function is used to calculate the all mathematical operations OR statistical operation of all function it is called as agg function 
# agg functon me sabhi mathematical operation ek sath execute kar sakte hai .... 
```

```python
import numpy as np 
import pandas as pd 
```

```python
# Pivot Tables 
```

```python
data = {
    "Date" : pd.date_range("2023-01-01",periods = 20),
    "Product" : ["A","B","C","D"] * 5,
    "Region" : ["East","West","North","South","East","West","North","South","East","West",
                 "North","South","East","West","North","South","East","West","North","South"],
    "Sales" : np.random.randint(100 , 1000, 20),
    "Units" : np.random.randint(10,100,20),
    "Rep" : ["John", "Mary", "Bob", "Alice","John", "Mary", "Bob", "Alice","John", "Mary",
             "Bob", "Alice","John", "Mary", "Bob", "Alice","John", "Mary", "Bob", "Alice",]
}

df = pd.DataFrame(data)

df["Month"] = df["Date"].dt.month_name()
df["Quarter"] = "Q" + df["Date"].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values = "Sales",index = "Region", columns = "Product",aggfunc= "median") # When you dont give the aggfunction so it takes the mean by default 
```

```python
pivot2 = pd.pivot_table(df, values = ["Sales","Units"] , index = "Region", columns = "Product")
pivot2
```

```python
pd.crosstab(df["Region"],df["Product"])  # Crosstab is for counting purpose 
```

```python
# Make a new column for episode count  

def extract_episodes(txt):
    data = ""
    check = False

    for i in txt:
        if i == "(":
            check = True

        elif i == ")":
            check = False
            return data

        elif check == True:
            data = data + i
    
    
```

```python
df["Episodes"] = df["Title"].apply(extract_episodes)
```

```python
df["Episodes"] = df["Episodes"].str.replace("eps","")  # for removing the eps word from the episodes because for aggregation function [ min , max] etc ..
df  # the episodes numbers is still in string 
```

```python
df["Episodes"] = df["Episodes"].astype(int)   # astype function is use to conversion of the data
```

```python
df
```

```python
# Question 2 = 
# make a new coloumn for time stamp

def extraction_time(txt): 
    check = False
    data = ""
    for i in range(len(txt)): 
        if txt[i] == ")": 
            for j in range(i+1, i + 20): 
                data += txt[j] 
            return data 
```

```python
df["Total Time"] = df["Title"].apply(extraction_time)
df
```

```python
# which anime hass the height score 
df[df["Score"] == df["Score"].max()]
```

```python
# Which anime has the episode count 
df[df["Episodes"] == df["Episodes"].max()]
```

```python
# Which is the longhest runtime anime 
df.sort_values('Episodes', ascending=False).head(1)
```

```python
# What is the capital of the country with least population 

df[df["population"] == df["population"].min()]["capital_city"]
```

```python
# give me top 5 countries with heigest democratic score  
df.sort_values("democracy_score",ascending = False , inplace= True)
df["country"].head()
```

```python
# How many region are there 
df["region"].value_counts().count()
```

```python
# How many countries lie in Rastern Europe region 
df[df["region"] == "Eastern Europe"]["country"]
```

```python
# who is the poltical leader of the 2nd highest populated country 

df[df["population"] == df["population"].nlargest(2).iloc[1]]["political_leader"]
```

```python
# how many countries are there whose political leaders are unknown 
df[df["political_leader"].isna()]["country"] #.count()
```

```python
# how many country have Republic in theire full name 
count = 0

def counting(txt): 
    global count 
    if "republic" in txt.lower(): 
        count = count + 1 
    return txt 

df['country_long'] = df["country_long"].apply(counting)
print(count)
```
